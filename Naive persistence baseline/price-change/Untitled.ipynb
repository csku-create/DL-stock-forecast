```python
from pathlib import Path
import pandas as pd
import numpy as np


# ============================================================
# Configuration
# ============================================================

BASE_DIR = Path(".")

MODELS = ["LSTM", "CNN", "LR"]

TICKERS = [
    "AAPL",
    "QCOM",
    "BNY",
    "C",
    "0097.KL",
    "0166.KL",
    "1023.KL",
    "5258.KL",
]


# ============================================================
# Required output columns
# ============================================================

OUTPUT_COLUMNS = [
    "Ticker",
    "Model",

    "train sample",

    # Train
    "train RMSE",
    "train MAE",
    "train R2",
    "train Accuracy",
    #"train Profit Index",
    #"train countp0",
    #"train Profit Index 1",
    #"train countp1",
    #"train Profit Index 2",
    #"train countp2",
    #"train Profit Index 3",
    #"train countp3",
    #"train Profit Index 4",
    #"train countp4",

    # Validate
    "validate sample",
    
    "validate RMSE",
    "validate MAE",
    "validate R2",
    "validate Accuracy",
    #"validate Profit Index",
    #"validate countp0",
    #"validate Profit Index 1",
    #"validate countp1",
    #"validate Profit Index 2",
    #"validate countp2",
    #"validate Profit Index 3",
    #"validate countp3",
    #"validate Profit Index 4",
    #"validate countp4",

    # Test
    "test sample",
    
    "test RMSE",
    "test MAE",
    "test R2",
    "test Accuracy",
    #"test Profit Index",
    #"test countp0",
    #"test Profit Index 1",
    #"test countp1",
    #"test Profit Index 2",
    #"test countp2",
    #"test Profit Index 3",
    #"test countp3",
    #"test Profit Index 4",
    #"test countp4",
]


OUTPUT_FILE = "all_models_experiment_results.csv"


# ============================================================
# Normalize column name
# ============================================================

def normalize_column_name(name):
    """
    Normalize a column name so that differences in:
        - uppercase/lowercase
        - spaces
        - underscores
        - hyphens

    do not affect matching.

    Example:
        Train_RMSE
        train RMSE
        train-rmse

    all become:
        trainrmse
    """

    return (
        str(name)
        .strip()
        .lower()
        .replace(" ", "")
        .replace("_", "")
        .replace("-", "")
    )


# ============================================================
# Detect ticker
# ============================================================

def get_ticker_from_path(file_path):
    """
    Examples:

        LSTM/AAPL_LSTM_experiment_results-1.csv
        CNN/QCOM_CNN_experiment_results-1.csv
        LR/0097.KL_LR_experiment_results-1.csv

    Returns:
        AAPL
        QCOM
        0097.KL
    """

    file_name = file_path.stem
    print(file_name)
    # --------------------------------------------------------
    # First search whether one of the required tickers
    # appears in the file name
    # --------------------------------------------------------

    for ticker in TICKERS:

        if file_name.upper().startswith(ticker.upper() + "_"):
            return ticker

    # --------------------------------------------------------
    # Try parent folder as backup
    # --------------------------------------------------------

    folder_name = file_path.parent.name
    print(folder_name)
    for ticker in TICKERS:

        if ticker.upper() in folder_name.upper():
            return ticker

    return None


# ============================================================
# Find matching source column
# ============================================================

def find_column(df, target_name):
    """
    Find a source column matching target_name.

    Example:

        target_name = "train RMSE"

    can match:

        train RMSE
        Train RMSE
        train_RMSE
        TRAIN_RMSE
        train-rmse
    """

    normalized_target = normalize_column_name(target_name)

    for col in df.columns:

        if normalize_column_name(col) == normalized_target:
            return col

    return None


# ============================================================
# Possible aliases
# ============================================================

COLUMN_ALIASES = {
    "train sample": [
        "Train Sample",
        "train sample",
    ],

    "validate sample": [
        "Validation Sample",
        "validation sample",
    ],

    "test sample": [
        "Test Sample",
        "test sample",
    ],

    # ========================================================
    # TRAIN
    # ========================================================

    "train RMSE": [
        "train RMSE",
        "Train RMSE",
        "train_RMSE",
    ],

    "train MAE": [
        "train MAE",
        "Train MAE",
        "train_MAE",
    ],

    "train R2": [
        "train R2",
        "Train R2",
        "train_R2",
        "train R²",
    ],

    "train Accuracy": [
        "train Accuracy",
        "Train Accuracy",
        "train_Accuracy",
    ],

    "train Profit Index": [
        "train Profit Index",
        "Train Profit Index",
        "train_Profit_Index",
        "train Profit Index 0",
        "train_Profit_Index_0",
    ],

    "train countp0": [
        "train countp0",
        "Train countp0",
        "train_countp0",
    ],

    "train Profit Index 1": [
        "train Profit Index 1",
        "Train Profit Index 1",
        "train_Profit_Index_1",
    ],

    "train countp1": [
        "train countp1",
        "Train countp1",
        "train_countp1",
    ],

    "train Profit Index 2": [
        "train Profit Index 2",
        "Train Profit Index 2",
        "train_Profit_Index_2",
    ],

    "train countp2": [
        "train countp2",
        "Train countp2",
        "train_countp2",
    ],

    "train Profit Index 3": [
        "train Profit Index 3",
        "Train Profit Index 3",
        "train_Profit_Index_3",
    ],

    "train countp3": [
        "train countp3",
        "Train countp3",
        "train_countp3",
    ],

    "train Profit Index 4": [
        "train Profit Index 4",
        "Train Profit Index 4",
        "train_Profit_Index_4",
    ],

    "train countp4": [
        "train countp4",
        "Train countp4",
        "train_countp4",
    ],


    # ========================================================
    # VALIDATE
    # ========================================================

    "validate RMSE": [
        "validate RMSE",
        "Validate RMSE",
        "validation RMSE",
        "Validation RMSE",
        "validate_RMSE",
        "validation_RMSE",
        "val RMSE",
        "val_RMSE",
    ],

    "validate MAE": [
        "validate MAE",
        "Validate MAE",
        "validation MAE",
        "Validation MAE",
        "validate_MAE",
        "validation_MAE",
        "val MAE",
        "val_MAE",
    ],

    "validate R2": [
        "validate R2",
        "Validate R2",
        "validation R2",
        "Validation R2",
        "validate_R2",
        "validation_R2",
        "val R2",
        "val_R2",
    ],

    "validate Accuracy": [
        "validate Accuracy",
        "Validate Accuracy",
        "validation Accuracy",
        "Validation Accuracy",
        "validate_Accuracy",
        "validation_Accuracy",
        "val Accuracy",
        "val_Accuracy",
    ],

    "validate Profit Index": [
        "validate Profit Index",
        "Validate Profit Index",
        "validation Profit Index",
        "Validation Profit Index",
        "validate_Profit_Index",
        "validation_Profit_Index",
        "val Profit Index",
        "val_Profit_Index",
        "validate Profit Index 0",
    ],

    "validate countp0": [
        "validate countp0",
        "validation countp0",
        "val countp0",
        "validate_countp0",
        "validation_countp0",
        "val_countp0",
    ],

    "validate Profit Index 1": [
        "validate Profit Index 1",
        "validation Profit Index 1",
        "val Profit Index 1",
        "validate_Profit_Index_1",
        "validation_Profit_Index_1",
        "val_Profit_Index_1",
    ],

    "validate countp1": [
        "validate countp1",
        "validation countp1",
        "val countp1",
        "validate_countp1",
        "validation_countp1",
        "val_countp1",
    ],

    "validate Profit Index 2": [
        "validate Profit Index 2",
        "validation Profit Index 2",
        "val Profit Index 2",
        "validate_Profit_Index_2",
        "validation_Profit_Index_2",
        "val_Profit_Index_2",
    ],

    "validate countp2": [
        "validate countp2",
        "validation countp2",
        "val countp2",
        "validate_countp2",
        "validation_countp2",
        "val_countp2",
    ],

    "validate Profit Index 3": [
        "validate Profit Index 3",
        "validation Profit Index 3",
        "val Profit Index 3",
        "validate_Profit_Index_3",
        "validation_Profit_Index_3",
        "val_Profit_Index_3",
    ],

    "validate countp3": [
        "validate countp3",
        "validation countp3",
        "val countp3",
        "validate_countp3",
        "validation_countp3",
        "val_countp3",
    ],

    "validate Profit Index 4": [
        "validate Profit Index 4",
        "validation Profit Index 4",
        "val Profit Index 4",
        "validate_Profit_Index_4",
        "validation_Profit_Index_4",
        "val_Profit_Index_4",
    ],

    "validate countp4": [
        "validate countp4",
        "validation countp4",
        "val countp4",
        "validate_countp4",
        "validation_countp4",
        "val_countp4",
    ],


    # ========================================================
    # TEST
    # ========================================================

    "test RMSE": [
        "test RMSE",
        "Test RMSE",
        "test_RMSE",
    ],

    "test MAE": [
        "test MAE",
        "Test MAE",
        "test_MAE",
    ],

    "test R2": [
        "test R2",
        "Test R2",
        "test_R2",
        "test R²",
    ],

    "test Accuracy": [
        "test Accuracy",
        "Test Accuracy",
        "test_Accuracy",
    ],

    "test Profit Index": [
        "test Profit Index",
        "Test Profit Index",
        "test_Profit_Index",
        "test Profit Index 0",
        "test_Profit_Index_0",
    ],

    "test countp0": [
        "test countp0",
        "Test countp0",
        "test_countp0",
    ],

    "test Profit Index 1": [
        "test Profit Index 1",
        "Test Profit Index 1",
        "test_Profit_Index_1",
    ],

    "test countp1": [
        "test countp1",
        "Test countp1",
        "test_countp1",
    ],

    "test Profit Index 2": [
        "test Profit Index 2",
        "Test Profit Index 2",
        "test_Profit_Index_2",
    ],

    "test countp2": [
        "test countp2",
        "Test countp2",
        "test_countp2",
    ],

    "test Profit Index 3": [
        "test Profit Index 3",
        "Test Profit Index 3",
        "test_Profit_Index_3",
    ],

    "test countp3": [
        "test countp3",
        "Test countp3",
        "test_countp3",
    ],

    "test Profit Index 4": [
        "test Profit Index 4",
        "Test Profit Index 4",
        "test_Profit_Index_4",
    ],

    "test countp4": [
        "test countp4",
        "Test countp4",
        "test_countp4",
    ],
}


# ============================================================
# Find a column using aliases
# ============================================================

def find_column_with_aliases(df, output_column):

    aliases = COLUMN_ALIASES.get(
        output_column,
        [output_column]
    )

    normalized_columns = {
        normalize_column_name(col): col
        for col in df.columns
    }

    for alias in aliases:

        key = normalize_column_name(alias)

        if key in normalized_columns:
            return normalized_columns[key]

    return None


# ============================================================
# Read all experiment result files
# ============================================================

rows = []


for model in MODELS:

    model_dir = BASE_DIR / f"{model}-ticker setting"

    files = list(
        model_dir.glob("*_NaivePersistence_experiment_results-2.csv")
    )

    print(
        f"\n{'=' * 60}\n"
        f"{model}: {len(files)} file(s) found\n"
        f"{'=' * 60}"
        f"{ model_dir}"
    )


    for file_path in files:

        print(f"\nReading: {file_path}")

        # ----------------------------------------------------
        # Determine ticker
        # ----------------------------------------------------

        ticker = get_ticker_from_path(file_path)

        if ticker is None:

            print(
                "  WARNING: Cannot determine ticker. "
                "Skipping."
            )

            continue


        # ----------------------------------------------------
        # Read CSV
        # ----------------------------------------------------

        try:

            df = pd.read_csv(file_path)

        except Exception as e:

            print(f"  ERROR reading CSV: {e}")

            continue


        if df.empty:

            print("  WARNING: CSV is empty.")

            continue


        # ----------------------------------------------------
        # Show original columns for debugging
        # ----------------------------------------------------

        print(f"  Ticker: {ticker}")
        print(f"  Columns: {list(df.columns)}")


        # ----------------------------------------------------
        # Create output row
        # ----------------------------------------------------

        row = {
            "Model": model,
            "Ticker": ticker,            
        }


        # ----------------------------------------------------
        # Extract each requested metric
        # ----------------------------------------------------

        metric_columns = OUTPUT_COLUMNS[2:]

        for output_col in metric_columns:

            source_col = find_column_with_aliases(
                df,
                output_col
            )

            if source_col is None:

                row[output_col] = np.nan

                print(
                    f"  WARNING: Cannot find "
                    f"'{output_col}'"
                )

            else:

                # Take first row
                row[output_col] = df[source_col].iloc[0]


        rows.append(row)


# ============================================================
# Create dataframe
# ============================================================

result = pd.DataFrame(
    rows,
    columns=OUTPUT_COLUMNS
)


# ============================================================
# Ensure all Model × Ticker combinations exist
#
# 3 models × 8 tickers = 24 rows
# ============================================================

expected = pd.MultiIndex.from_product(
    [
        MODELS,
        TICKERS,        
    ],
    names=[
        "Model",
        "Ticker",        
    ]
).to_frame(index=False)


result = expected.merge(
    result,
    on=[
        "Model",
        "Ticker",        
    ],
    how="left"
)


# ============================================================
# Sort ticker and model
# ============================================================

result["Ticker"] = pd.Categorical(
    result["Ticker"],
    categories=TICKERS,
    ordered=True
)

result["Model"] = pd.Categorical(
    result["Model"],
    categories=MODELS,
    ordered=True
)


result = (
    result
    .sort_values(
        [
            "Model",
            "Ticker",            
        ]
    )
    .reset_index(drop=True)
)


# ============================================================
# Convert categorical columns back to strings
# ============================================================

result["Ticker"] = result["Ticker"].astype(str)
result["Model"] = result["Model"].astype(str)


# ============================================================
# Final column order
# ============================================================

result = result[OUTPUT_COLUMNS]


# ============================================================
# Save CSV
# ============================================================

result.to_csv(
    OUTPUT_FILE,
    index=False,
    na_rep="na"
)


# ============================================================
# Display result
# ============================================================

print("\n")
print("=" * 80)
print("Completed")
print("=" * 80)

print(result.to_string(index=False))

print("\n" + "=" * 80)
print(f"Saved to: {OUTPUT_FILE}")
print(f"Total rows: {len(result)}")
print("=" * 80)
```
